```python
import pandas as pd
from scipy.stats import wilcoxon
from scipy.stats import f_oneway
import matplotlib.pyplot as plt
import numpy as np
```

Mūsų failų transkripcijos

```python
bazinis = pd.read_csv("results_bazinio.csv")
trecdalis = pd.read_csv("results_trecdalio.csv")
duomenu_atranka = pd.read_csv("results_further_liepa2.csv")
```

Testinės aibės 1000 failų transkripcijos

```python
bazinis_validacijos = pd.read_csv("validation_results_base.csv")
trecdalis_validacijos = pd.read_csv("validation_results_trecdalio.csv")
duomenu_atranka_validacijos = pd.read_csv("validation_results.csv")
```

```python
bazinis_medicinos = pd.read_csv("wer_cer_bazinis_med.csv")
trecdalis_medicinos = pd.read_csv("wer_cer_results_med_trecdalio.csv")
duomenu_atranka_medicinos = pd.read_csv("wer_cer_results_med_duomenu_atrankos.csv")
```

Mūsų duomenų Bazinis VS Dalinai apmokytas

```python
stat_b_t, p_b_t = wilcoxon(bazinis["wer"], trecdalis["wer"])

print("A vs B")
print("stat:", stat_b_t)
print("p-value:", p_b_t)
```

```python
stat_b_t_cer, p_b_t_cer = wilcoxon(bazinis["cer"], trecdalis["cer"])

print("A vs B")
print("stat:", stat_b_t_cer)
print("p-value:", p_b_t_cer)
```

```python
stat_b_d_cer, p_b_d_cer = wilcoxon(bazinis["cer"], duomenu_atranka["cer"])

print("A vs B")
print("stat:", stat_b_d_cer)
print("p-value:", p_b_d_cer)
```

```python
stat_t_d, p_t_d = wilcoxon(trecdalis["wer"], duomenu_atranka["wer"])

print("A vs B")
print("stat:", stat_t_d)
print("p-value:", p_t_d)
```

```python
stat_t_d_cer, p_t_d_cer = wilcoxon(trecdalis["cer"], duomenu_atranka["cer"])

print("A vs B")
print("stat:", stat_t_d_cer)
print("p-value:", p_t_d_cer)
```

```python
stat_t_d_cer_2, p_t_d_cer_2 = wilcoxon(trecdalis["cer"], duomenu_atranka["cer"], alternative="greater")

print("A vs B")
print("stat:", stat_t_d_cer_2)
print("p-value:", p_t_d_cer_2)
```

Mūsų įrašų transkripcijų palyginimo grafikai


```python
files = bazinis["file"]
x = np.arange(len(files))
width = 0.25

plt.figure(figsize=(14,6))

plt.bar(x - width, bazinis["wer"], width=width, label="Bazinis modelis")
plt.bar(x, trecdalis["wer"], width=width, label="Dalimi duomenų apmokytas modelis")
plt.bar(x + width, duomenu_atranka["wer"], width=width, label="Atrinktais duomenimis apmokytas modelis")

plt.xticks(x, files, rotation=90)
plt.xlabel("Įrašas")
plt.ylabel("WER")
plt.title("Mūsų duomenų rinkinio transkripcijų WER palyginimas tarp modelių")
plt.legend()

plt.tight_layout()
plt.savefig("WER_palyginimas_musu_duomenim.pdf")
plt.show()

```

```python
plt.figure(figsize=(14,6))

plt.bar(x - width, bazinis["cer"], width=width, label="Bazinis modelis")
plt.bar(x, trecdalis["cer"], width=width, label="Dalimi duomenų apmokytas modelis")
plt.bar(x + width, duomenu_atranka["cer"], width=width, label="Atrinktais duomenimis apmokytas modelis")

plt.xticks(x, files, rotation=90)
plt.xlabel("Įrašas")
plt.ylabel("CER")
plt.title("Mūsų duomenų rinkinio transkripcijų CER palyginimas tarp modelių")
plt.legend()

plt.tight_layout()
plt.savefig("CER_palyginimas_musu_duomenim.pdf")
plt.show()
```
